```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = {
    "Hours_Studied": [2, 5, 3, 7, 1, 6, 4, 8, 5, 3],
    "Attendance": [60, 90, 75, 95,50,88,80,98,85,70],
    "Previous_Score": [45,70,55,82,40,78,65,90,72,58],
    "Final_Score": [42,76,58,88,35,82,68,94,79,60]
}
df = pd.DataFrame(data)
print(df.head())
print("Shape:",df.shape)
print(df.isnull().sum())

sns.scatterplot(data=df, x="Hours_Studied", y="Final_Score")
plt.title("Study Hours vs Final Score")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = {
    "Product": ["Laptop", "Mouse", "Keyboard", "Monitor", "Headset", "Webcam", "Charger", "USB_Cable", "Tablet", "Speaker"],
    "Units_Sold": [15, 85, 50, 22, 40, 35, 110, 150, 18, 500],
    "Price": [800, 25, 45, 200, 60, 50, 20, 10, 300, 75],
    "Revenue": [12000, 2125, 2250, 4400, 2400, 1750, 2200, None, 5400, 37500]
}

df = pd.DataFrame(data)

print(df.head())
print(df.shape)

print(df.dtypes)
print(df.isnull().sum())

sns.barplot(data=df, x="Product", y="Units_Sold")
plt.title("Units Sold per Product")
plt.xticks(rotation=45)
plt.show()
```
