```python

import os
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression

csv_file = "student.csv"

df = pd.read_csv(csv_file)

label_map = {"Fail": 0, "Pass": 1}
inv_label_map = {0: "Fail", 1: "Pass"}

features = [
    "Study Hours",
    "Attendance (%)",
    "Previous Marks",
    "Final Marks"
]

X = df[features]

y = (
    (df["Attendance (%)"] > 75) &
    (df["Final Marks"] > 35)
).astype(int)

lr_result_model = LinearRegression()
lr_result_model.fit(X, y)

study_hours = float(input("Enter Study Hours: "))
attendance = float(input("Enter Attendance (%): "))
previous_marks = float(input("Enter Previous Marks: "))
final_marks = float(input("Enter Final Marks: "))

new_X = pd.DataFrame(
    [[study_hours, attendance, previous_marks, final_marks]],
    columns=features
)

raw_pred = lr_result_model.predict(new_X)[0]

rounded_pred = int(np.clip(np.rint(raw_pred), 0, 1))

predicted_result = inv_label_map[rounded_pred]

print("\nPredicted Result:", predicted_result)

new_row = pd.DataFrame(
    [[study_hours, attendance, previous_marks,
      final_marks, predicted_result]],
    columns=[
        "Study Hours",
        "Attendance (%)",
        "Previous Marks",
        "Final Marks",
        "Result"
    ]
)

new_row.to_csv(
    csv_file,
    mode="a",
    index=False,
    header=False
)

print("\nNew Student Record:")
print(new_row)
```

```python
import os
import pandas as pd
from sklearn.linear_model import LogisticRegression

csv_file = "student.csv"
X_train = pd.DataFrame({
    "Attendance_Status": [0, 0, 1, 1],
    "Marks_Status": [0, 1, 0, 1]
})
y_train = [0, 0, 0, 1]
log_model = LogisticRegression(C=1000, max_iter=2000)
log_model.fit(X_train, y_train)

study_hours = float(input("Enter Study Hours: "))
attendance = float(input("Enter Attendance (%): "))
previous_marks = float(input("Enter Previous Marks: "))
final_marks = float(input("Enter Final Marks: "))

new_X = pd.DataFrame({
    "Attendance_Status": [int(attendance > 75)],
    "Marks_Status": [int(final_marks > 35)]
})

pred_code = int(log_model.predict(new_X)[0])

class_lookup = {
    0: "Fail",
    1: "Pass"
}

predicted_result = class_lookup[pred_code]

print("\nPredicted Result:", predicted_result)

new_row = pd.DataFrame({
    "Study Hours": [study_hours],
    "Attendance (%)": [attendance],
    "Previous Marks": [previous_marks],
    "Final Marks": [final_marks],
    "Result": [predicted_result]
})

file_exists = os.path.exists(csv_file) and os.path.getsize(csv_file) > 0

new_row.to_csv(
    csv_file,
    mode="a",
    index=False,
    header=not file_exists
)

updated_df = pd.read_csv(csv_file)

print("\nNew Student Record:")
print(updated_df.tail(1).to_string(index=False))
```

```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    f1_score,
    mean_absolute_error,
    mean_squared_error,
    precision_score,
    r2_score,
    recall_score,
)
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

df = pd.read_csv("student.csv")
df["Result"] = df["Result"].astype(str).str.strip()

features = ["Study Hours", "Attendance (%)", "Previous Marks"]
X = df[features]

y_reg = df["Final Marks"]
X_train_r, X_test_r, y_train_r, y_test_r = train_test_split(
    X, y_reg, test_size=0.2, random_state=42
)

reg_model = LinearRegression()
reg_model.fit(X_train_r, y_train_r)
y_pred_r = reg_model.predict(X_test_r)

mae = mean_absolute_error(y_test_r, y_pred_r)
mse = mean_squared_error(y_test_r, y_pred_r)
rmse = np.sqrt(mse)
r2 = r2_score(y_test_r, y_pred_r)

y_clf = df["Result"].map({"Fail": 0, "Pass": 1})
X_train_c, X_test_c, y_train_c, y_test_c = train_test_split(
    X, y_clf, test_size=0.2, random_state=42, stratify=y_clf
)

clf_model = make_pipeline(StandardScaler(), LogisticRegression(max_iter=2000))
clf_model.fit(X_train_c, y_train_c)
y_pred_c = clf_model.predict(X_test_c)

acc = accuracy_score(y_test_c, y_pred_c)
prec = precision_score(y_test_c, y_pred_c, average="weighted", zero_division=0)
rec = recall_score(y_test_c, y_pred_c, average="weighted", zero_division=0)
f1 = f1_score(y_test_c, y_pred_c, average="weighted", zero_division=0)

print("=== Regression Model  (LinearRegression) ===")
print(f"MAE : {mae:.4f}")
print(f"MSE : {mse:.4f}")
print(f"RMSE: {rmse:.4f}")
print(f"R2  : {r2:.4f}")

print("\n=== Classification Model (LogisticRegression) ===")
print(f"Accuracy : {acc:.4f}")
print(f"Precision: {prec:.4f}")
print(f"Recall   : {rec:.4f}")
print(f"F1 Score : {f1:.4f}")

comparison_df = pd.DataFrame({
    "Comparison Point": [
        "Algorithm used",
        "Target variable",
        "Type of prediction",
        "Evaluation metrics",
        "Real-world application",
    ],
    "Regression Model ": [
        "LinearRegression",
        "Final Marks",
        "Continuous numerical value",
        "MAE, MSE, RMSE, R2",
        "Forecasting student exam marks",
    ],
    "Classification Model ": [
        "LogisticRegression (with StandardScaler)",
        "Result (Pass/Fail encoded as 1/0)",
        "Discrete class label (Pass or Fail)",
        "Accuracy, Precision, Recall, F1 Score",
        "Identifying students likely to Pass or Fail for early support",
    ],
})

print(comparison_df.to_string(index=False))
```
